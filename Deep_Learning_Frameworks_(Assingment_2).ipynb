## Q.1 - What is TensorFlow 2.0, and how is it different from TensorFlow 1.x?

Ans - ### TensorFlow 2.0 vs. TensorFlow 1.x

TensorFlow 2.0 is a major upgrade from TensorFlow 1.x, introduced in September 2019. It focuses on simplicity, ease of use, and improved performance. This update makes deep learning model development more intuitive, efficient, and ready for production.  

### Key Differences Between TensorFlow 2.0 and 1.x

#### 1. Eager Execution (Enabled by Default)
TensorFlow 1.x required defining a static computation graph and executing it within a `tf.Session()`, making debugging complex. TensorFlow 2.0 enables eager execution by default, allowing operations to be executed immediately, similar to standard Python code. This makes debugging and experimentation easier.  

#### 2. Keras as the Central API
In TensorFlow 1.x, model building involved multiple approaches (`tf.layers`, `tf.contrib`, `Estimator API`), making it confusing. TensorFlow 2.0 standardizes `tf.keras` as the primary high-level API for defining and training models, promoting a more Pythonic experience.  

#### 3. Simplified and Cleaned-up API  
TensorFlow 2.0 removes redundant or deprecated APIs (`tf.app`, `tf.flags`, `tf.logging`) and consolidates multiple methods to streamline development. Many low-level operations are now simplified for easier implementation.  

#### 4. Enhanced Dataset Handling with `tf.data`  
In TensorFlow 1.x, data input relied heavily on `feed_dict`, which was inefficient. TensorFlow 2.0 encourages using the `tf.data` API, allowing efficient data loading, preprocessing, and batch management for large datasets.  

#### 5. Improved Distributed Training
Previously, multi-GPU and TPU training required complex configurations. TensorFlow 2.0 introduces `tf.distribute.Strategy`, which simplifies training across multiple devices without modifying core code.  

#### 6. Better Deployment and Production Readiness  
TensorFlow 2.0 improves integration with TensorFlow Serving, TensorFlow Lite, and TensorFlow.js, making it easier to deploy models across various platforms (mobile, web, cloud).

## Q.3 - What is the primary function of the tf.function in TensorFlow 2.0?
Ans - ### Primary Function of `tf.function` in TensorFlow 2.0

In TensorFlow 2.0, `tf.function` is used to convert Python functions into TensorFlow’s computational graphs (graph mode). This improves performance by optimizing execution speed, reducing overhead, and enabling efficient execution on GPUs and TPUs.  

#### Key Benefits:  
1. Performance Optimization – Graph execution is faster than eager execution due to compiler optimizations.  
2. Automatic Graph Compilation – Converts Python functions into TensorFlow graphs for efficient execution.  
3. Auto-differentiation Support – Works with `tf.GradientTape` to compute gradients in deep learning models.  
4. Seamless Deployment – Optimized graphs can be saved and deployed using TensorFlow Serving, Lite, and JS.  

#### **Example Usage:**  
```python
import tensorflow as tf

@tf.function
def multiply(x, y):
    return x * y

a = tf.constant(3)
b = tf.constant(4)
print(multiply(a, b))  
```
Here, `tf.function` automatically compiles `multiply(x, y)` into a computational graph for faster execution.  

#### When to Use `tf.function`?  
- For high-performance model training and inference.  
- When deploying models in production.  
- Avoid when debugging, as eager execution provides clearer error messages.

## Q.5 -  How do you create a neural network using TensorFlow 2.0?
Ans -

````python
import tensorflow as tf
from tensorflow.keras import layers, models
from tensorflow.keras.datasets import mnist
import numpy as np

(X_train, y_train), (X_test, y_test) = mnist.load_data()

X_train = X_train.reshape(-1, 784).astype('float32') / 255.0
X_test = X_test.reshape(-1, 784).astype('float32') / 255.0

model = models.Sequential([
    layers.Dense(128, activation='relu', input_shape=(784,)),
    layers.Dense(64, activation='relu'),
    layers.Dense(10, activation='softmax')
])

model.compile(optimizer='adam',
              loss='sparse_categorical_crossentropy',
              metrics=['accuracy'])

model.fit(X_train, y_train, epochs=10, batch_size=32, validation_split=0.2)

test_loss, test_acc = model.evaluate(X_test, y_test)
print(f"Test Accuracy: {test_acc:.4f}")
````

## Q. 10 - What is the difference between TensorFlow and PyTorch?
Ans -

### Theoretical Difference Between TensorFlow and PyTorch  

TensorFlow and PyTorch are the two leading deep learning frameworks, both widely used for building and training neural networks. While both frameworks provide similar functionality, they differ in execution style, ease of use, deployment capabilities, and ecosystem support.  

---

### 1. Execution Style: Static vs. Dynamic Computation Graphs  
The fundamental difference between TensorFlow and PyTorch lies in how they execute computational graphs:  

- TensorFlow primarily uses a static computation graph (Graph Mode), meaning the computation is defined first and executed later. This allows for graph optimization and improved performance, especially in large-scale applications.  
- PyTorch, on the other hand, employs a dynamic computation graph (Eager Mode), meaning computations are executed immediately. This makes it more intuitive and easier to debug, as errors can be spotted in real-time.  

While TensorFlow 2.0 introduced Eager Execution to improve flexibility, PyTorch remains the preferred choice for research due to its dynamic nature.  


### 2. Ease of Use and Debugging  
- PyTorch is more Pythonic, making it easier to write, debug, and experiment with models. Its intuitive syntax makes it popular in academia and research.  
- TensorFlow, particularly in version 1.x, was more complex, requiring session-based execution. However, TensorFlow 2.0 introduced `tf.keras` for a higher-level API, improving usability. Debugging in TensorFlow can still be more challenging due to its static graph execution.  


### 3. Deployment and Production
- TensorFlow has a well-established deployment pipeline with TensorFlow Serving, TensorFlow Lite, and TensorFlow.js, making it a preferred choice for large-scale production and mobile applications.  
- PyTorch has improved in deployment with TorchScript and ONNX, but it still lags behind TensorFlow in terms of industrial scalability.  


### 4. Ecosystem and Community Support  
- TensorFlow has a larger ecosystem, including TensorBoard for visualization and TPU support for high-performance training.  
- PyTorch is widely used in research due to its flexibility and strong support from the academic community.

## Q. 12 - What is the basic structure of a PyTorch neural network?
Ans -
In PyTorch, a basic neural network is typically built using the torch.nn module. The core structure of a PyTorch neural network consists of the following components:

2. Define the Neural Network Model
The model is created as a class that inherits from nn.Module. The key methods are:

__init__(): Defines the network layers.
forward(x): Defines the forward pass of the network.

3. Instantiate the Model
4. Define Loss Function and Optimizer
5. Training Loop
6. Evaluation
After training, you can evaluate the model using:


````python
import torch
import torch.nn as nn
import torch.optim as optim
import torch.nn.functional as F
````

````python
class SimpleNN(nn.Module):
    def __init__(self):
        super(SimpleNN, self).__init__()
        self.fc1 = nn.Linear(10, 20)
        self.fc2 = nn.Linear(20, 1)

    def forward(self, x):
        x = F.relu(self.fc1(x))
        x = self.fc2(x)
        return x
````

````python
model = SimpleNN()
````

````python
criterion = nn.MSELoss()
optimizer = optim.SGD(model.parameters(), lr=0.01)
````

````python
for epoch in range(100):
    optimizer.zero_grad()
    inputs = torch.randn(5, 10)
    targets = torch.randn(5, 1)

    outputs = model(inputs)
    loss = criterion(outputs, targets)
    loss.backward()
    optimizer.step()

    print(f"Epoch {epoch+1}, Loss: {loss.item():.4f}")
````

````python
with torch.no_grad():
    test_input = torch.randn(1, 10)
    prediction = model(test_input)
    print("Prediction:", prediction)
````

## Q. 17 - What is the difference between torch.nn.Module and torch.nn.Sequential in PyTorch?

### Difference Between `torch.nn.Module` and `torch.nn.Sequential` in PyTorch

Both `torch.nn.Module` and `torch.nn.Sequential` are used to define neural network models in PyTorch, but they serve different purposes and provide different levels of flexibility.



### 1. `torch.nn.Module` (Base Class for Models)
- Definition: `torch.nn.Module` is the base class for all PyTorch neural network models.
- Purpose: It allows defining complex models with custom forward passes and submodules.
- Flexibility: Provides full control over the forward pass, enabling conditional logic, multiple inputs/outputs, and customized layers.
- Usage: Recommended for complex architectures such as ResNet, transformers, and GANs.

Example (Custom Neural Network using `nn.Module`):
```python
import torch
import torch.nn as nn

class CustomModel(nn.Module):
    def __init__(self):
        super(CustomModel, self).__init__()
        self.fc1 = nn.Linear(10, 5)
        self.relu = nn.ReLU()
        self.fc2 = nn.Linear(5, 2)

    def forward(self, x):
        x = self.fc1(x)
        x = self.relu(x)
        x = self.fc2(x)
        return x

model = CustomModel()
print(model)
```



### 2. `torch.nn.Sequential` (Simplified Model Definition)
- Definition: `torch.nn.Sequential` is a container that stacks layers in a sequential order.
- Purpose: Provides a quick and simple way to define feedforward models.
- Flexibility: Limited compared to `nn.Module`, as it enforces a strict layer order and does not allow custom forward logic.
- Usage: Ideal for straightforward feedforward architectures like CNNs and MLPs.

Example (Same model using `nn.Sequential`):
```python
model = nn.Sequential(
    nn.Linear(10, 5),
    nn.ReLU(),
    nn.Linear(5, 2)
)
print(model)
```


### Key Differences
| Feature | `torch.nn.Module` | `torch.nn.Sequential` |
|---------|------------------|----------------------|
| **Customization** | Allows custom forward pass with complex logic | Fixed layer order, no custom logic |
| **Flexibility** | Can handle multiple inputs, loops, and conditions | Simple feedforward models only |
| **Ease of Use** | Requires defining `forward()` manually | Automatically infers forward pass |
| **Best For** | Complex models (CNNs, RNNs, transformers) | Simple MLPs, CNNs, feedforward networks |

## Q.4 - How can you visualize the training progress using TensorFlow and Matplotlib?

````python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
import numpy as np
import matplotlib.pyplot as plt


X_train = np.random.rand(200, 10)
y_train = np.random.randint(0, 2, size=(200,))  # Binary labels


model = keras.Sequential([
    layers.Dense(16, activation='relu', input_shape=(10,)),
    layers.Dense(1, activation='sigmoid')
])


model.compile(optimizer='adam',
              loss='binary_crossentropy',
              metrics=['accuracy'])


history = model.fit(X_train, y_train, epochs=20, batch_size=10, validation_split=0.2)
````

````python

train_loss = history.history['loss']
val_loss = history.history['val_loss']
train_acc = history.history['accuracy']
val_acc = history.history['val_accuracy']
epochs = range(1, len(train_loss) + 1)


plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.plot(epochs, train_loss, 'bo-', label='Training Loss')
plt.plot(epochs, val_loss, 'r*-', label='Validation Loss')
plt.title('Training and Validation Loss')
plt.xlabel('Epochs')
plt.ylabel('Loss')
plt.legend()


plt.subplot(1, 2, 2)
plt.plot(epochs, train_acc, 'bo-', label='Training Accuracy')
plt.plot(epochs, val_acc, 'r*-', label='Validation Accuracy')
plt.title('Training and Validation Accuracy')
plt.xlabel('Epochs')
plt.ylabel('Accuracy')
plt.legend()

plt.show()
````

````python
import torch
import torch.nn as nn
import torch.optim as optim


class SimpleNN(nn.Module):
    def __init__(self):
        super().__init__()
        self.layer1 = nn.Linear(10, 16)
        self.layer2 = nn.Linear(16, 1)

    def forward(self, x):
        x = torch.relu(self.layer1(x))
        return torch.sigmoid(self.layer2(x))


model = SimpleNN()
criterion = nn.BCELoss()
optimizer = optim.Adam(model.parameters(), lr=0.01)


X_train = torch.rand(100, 10)
y_train = torch.randint(0, 2, (100, 1), dtype=torch.float32)


for epoch in range(5):
    optimizer.zero_grad()
    output = model(X_train)
    loss = criterion(output, y_train)
    loss.backward()
    optimizer.step()
    print(f"Epoch {epoch+1}, Loss: {loss.item():.4f}")


X_test = torch.rand(5, 10)
print("Predictions:\n", model(X_test).detach().numpy())
````

## Q.8 - How do you implement a custom loss function in PyTorch?

````python
import torch
import torch.nn as nn

# Custom loss function: Mean Absolute Error (MAE)
def custom_mae_loss(y_pred, y_true):
    return torch.mean(torch.abs(y_pred - y_true))

# Example usage
y_true = torch.tensor([1.0, 0.0, 1.0])
y_pred = torch.tensor([0.9, 0.2, 0.8])

loss = custom_mae_loss(y_pred, y_true)
print("Custom Loss:", loss.item())
````

## Q.9 - How do you save and load a TensorFlow model?

````python
import tensorflow as tf

def save_model(model, filename="my_model.keras"):
    model.save(filename)
    print(f"Model saved as {filename}")
````

````python
def load_model(filename="my_model.keras"):
    model = tf.keras.models.load_model(filename)
    print(f"Model loaded from {filename}")
    return model
````
