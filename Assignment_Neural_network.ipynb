 Q. 3  What is the mathematical structure of a neural network!

Ans -  ### Mathematical Structure of a Neural Network -  

A neural network is a computational model that maps inputs to outputs using layers of interconnected neurons. Mathematically, it consists of weighted sums, activation functions, and optimization techniques** to learn patterns from data.  



#### 1. Layers and Neurons
A neural network consists of:  
- Input layer**: Takes input features \( X \).  
- Hidden layers: Perform computations using weights \( W \) and biases \( b \).  
- Output layer: Produces the final prediction.  

For a neuron in layer \( l \):  

\[
Z^{(l)} = W^{(l)} A^{(l-1)} + b^{(l)}
\]

\[
A^{(l)} = f(Z^{(l)})
\]

where:  
- \( A^{(l-1)} \) is the previous layer’s activation,  
- \( W^{(l)} \) and \( b^{(l)} \) are the weights and biases,  
- \( f \) is the activation function (e.g., ReLU, Sigmoid, Tanh).  



#### 2. Activation Functions
To introduce non-linearity, common functions include:  
- ReLU: \( f(x) = \max(0, x) \)  
- Sigmoid: \( f(x) = \frac{1}{1 + e^{-x}} \)  
- Softmax: Converts outputs into probabilities for classification.  



#### 3. Cost Function & Backpropagation**  
The network optimizes a cost function:  
- Mean Squared Error (MSE)** for regression  
- Cross-Entropy Loss** for classification  

Using gradient descent, weights are updated as:  

\[
W = W - \alpha \frac{\partial J}{\partial W}
\]

where \( \alpha \) is the learning rate, and \( J \) is the loss function.

#### 3. Sigmoid (Logistic Function) -
\[
f(x) = \frac{1}{1 + e^{-x}}

#### 4. Tanh (Hyperbolic Tangent Function) -
\[
f(x) = \frac{e^x - e^{-x}}{e^x + e^{-x}}

- Usage : Alternative to sigmoid, used in RNNs.

- Cons : Still suffers from vanishing gradients.  



#### 5. Softmax -  
\[
f(x_i) = \frac{e^{x_i}}{\sum e^{x_j}}
\]  
- Usage : Multi-class classification (last layer).  
- Pros : Converts outputs into probabilities.  
- Cons : Computationally expensive for large outputs.

A loss function is a mathematical function that measures the difference between a neural network’s predicted output and the actual target value. It quantifies the model’s error and guides the learning process.  



### * Loss Function Crucial -
1. Guides Learning – It provides feedback on how well the model is performing.

  \[
  J = \frac{1}{m} \sum (y_i - \hat{y}_i)^2
  \]

  \[

  \]  

A good loss function helps neural networks converge efficiently and generalize well to unseen data.

Ans -

### Common Types of Loss Functions in Neural Networks -

Loss functions measure how well a neural network’s predictions match actual values. The choice of a loss function depends on the task: regression or classification.  



#### * 1. Loss Functions for Regression
Used when the output is continuous.  

 *Mean Squared Error (MSE)
\[
J = \frac{1}{m} \sum (y_i - \hat{y}_i)^2
\]  
- Penalizes large errors more.  
- Sensitive to outliers.  

 * Mean Absolute Error (MAE)
\[
J = \frac{1}{m} \sum |y_i - \hat{y}_i|
\]  
- Treats all errors equally.  
- More robust to outliers than MSE.  

 * Huber Loss  
- Combines MSE (for small errors) and MAE (for large errors).  
- Useful when data has outliers.  



#### *2. Loss Functions for Classification  
Used when predicting categorical labels.  

 * Binary Cross-Entropy (for binary classification)  
\[
J = - \frac{1}{m} \sum [y_i \log(\hat{y}_i) + (1 - y_i) \log(1 - \hat{y}_i)]
\]  
- Works well for yes/no problems (e.g., spam detection).  

 * Categorical Cross-Entropy (for multi-class classification)  
\[
J = - \sum y_i \log(\hat{y}_i)
\]  
- Used when classes are one-hot encoded.  

 * Sparse Categorical Cross-Entropy
- Similar to categorical cross-entropy but works with integer labels instead of one-hot encoding.  

 *Hinge Loss  
\[
J = \sum \max(0, 1 - y_i \hat{y}_i)
\]  
- Used for Support Vector Machines (SVMs).  


Q. 9 - How does a neural network learn?

### How Does a Neural Network Learn -

A neural network learns by adjusting its weights and biases to minimize the error between predictions and actual outputs. This process involves three main steps  


### 1. Forward Propagation
- Input data passes through the network, layer by layer.  
- Each neuron applies a weighted sum and an activation function :  
  \[
  Z = WX + b
  \]
  \[
  A = f(Z)
  \]  
- The output layer produces the final prediction.  



### 2. Loss Calculation
- A loss function measures the difference between predicted and actual values.  
- Common loss functions:  
  - MSE (Mean Squared Error) for regression.  
  - Cross-Entropy Loss for classification.  



### 3. Backpropagation & Weight Update  
- The backpropagation algorithm computes the gradient of the loss function with respect to each weight using chain rule differentiation.  
- Gradient Descent updates weights:  
  \[
  W = W - \alpha \frac{\partial J}{\partial W}
  \]  
  where \( \alpha \) is the learning rate.  
- Popular optimizers like Adam and SGD improve learning efficiency.

### Forward and Backward Propagation in a Neural Network -

A neural network learns by passing data through the network (forward propagation) and then adjusting its weights to reduce errors (backward propagation).  



### * 1. Forward Propagation (Making a Prediction)  
Think of this as a student answering a question :  
1. The input (like a math problem) is given to the network.  
2. Each layer processes the data using weights, biases, and an activation function.  
3. The final layer gives an output (like the student’s answer).  

Mathematically, each neuron computes:  
\[
Z = WX + b
\]
\[
A = f(Z)
\]  
where \( W \) are weights, \( b \) is bias, and \( f \) is an activation function.  

If the output is wrong, we need to fix the network’s "understanding".  



### * 2. Backward Propagation (Learning from Mistakes)   
This is like a teacher correcting the student:  
1. Compare the output with the correct answer using a loss function.  
2. Compute the error (gradient) and figure out how much each weight contributed to the mistake.  
3. Adjust the weights using gradient descent:  
   \[
   W = W - \alpha \frac{\partial J}{\partial W}
   \]  
   where \( \alpha \) is the learning rate.  





Q. 13 - What is weight initialization, and how does it impact training?

Ans -
###  Weight Initialization -
Weight initialization is the process of setting the initial values of a neural network’s weights before training begins. Proper initialization is crucial because it affects how quickly and effectively a model learns.  


### * Initialization Impact Training -
1. Prevents Vanishing/Exploding Gradients  
   - If weights are too small, gradients shrink (vanishing gradient problem), slowing learning.  
   - If weights are too large, gradients explode, making training unstable.  

2. Speeds Up Convergence  
   - Well-initialized weights help the network reach optimal solutions **faster**.  
   - Poor initialization can cause slow or failed learning.  

3. Avoids Symmetry Problems
   - If all weights are initialized the same, neurons will learn **identically**, reducing the network’s ability to learn diverse features.  


### Common Weight Initialization Techniques  
Zero Initialization (Not recommended)  
- Sets all weights to zero.  
- Causes all neurons to learn the same thing (symmetry problem).  

Random Initialization
- Assigns small random values to weights.  
- Can still lead to unstable gradients.  

Xavier/Glorot Initialization (For Sigmoid/Tanh activations)  
\[
W \sim \mathcal{N}(0, \frac{1}{n})
\]  
- Balances variance across layers.  

He Initialization (For ReLU activations)  
\[
W \sim \mathcal{N}(0, \frac{2}{n})
\]  
- Prevents vanishing gradients in deep networks.  


Q. 14 - What is the vanishing gradient problem in deep learning?
Ans -
### Vanishing Gradient Problem in Deep Learning -

The vanishing gradient problem occurs when gradients become too small during backpropagation, causing early layers in deep neural networks to learn very slowly or not at all. This prevents effective training and leads to poor performance.  


### Why Does It Happen -  
1. Activation Functions – Functions like Sigmoid and Tanh squash inputs into small ranges (\(0\) to \(1\) for Sigmoid, \(-1\) to \(1\) for Tanh). Their derivatives are small for large or small inputs, making gradients shrink.  
2. Deep Networks – As gradients are propagated backward through multiple layers, they become smaller and smaller, making weight updates negligible.  

Mathematically, if gradients approach zero, weight updates:  
\[
W = W - \alpha \frac{\partial J}{\partial W}
\]  
become very small, slowing down learning.  


### *Effects of Vanishing Gradients
- Slow learning in early layers  
- Difficulty training deep networks
- Poor convergence and accuracy


### *Solutions to Vanishing Gradients  
ReLU Activation Function – Avoids small gradients by using \( f(x) = \max(0, x) \).  
He Initialization – Sets better starting weights.  
Batch Normalization – Stabilizes activations and gradients.  
Residual Connections (ResNets) – Help gradients flow freely.  


Q. 15 - What is the exploding gradient problem?
Ans -
### Exploding Gradient Problem -  

The exploding gradient problem occurs when gradients become excessively large during backpropagation, leading to unstable training and diverging weight updates. This makes the neural network unable to learn properly.  

### Why Does It Happen -  
1. Deep Networks – In very deep networks, gradients are multiplied layer by layer, sometimes growing exponentially.  
2. Large Weight Updates – When gradients are too large, weight updates:  
   \[
   W = W - \alpha \frac{\partial J}{\partial W}
   \]  
   cause drastic changes, leading to instability.  

### Effects of Exploding Gradients
- Diverging loss (NaN values in training)  
- Unstable model behavior  
- Difficult optimization  

### Solutions  
Gradient Clipping – Limits maximum gradient values.  
Proper Weight Initialization – Techniques like He Initialization prevent large values.  
Smaller Learning Rates – Reduces drastic weight updates.  
Batch Normalization – Stabilizes training by normalizing activations.

Q, 1 - How do you create a simple perceptron for basic binary classification?

```python
import numpy as np

def step(z):
    return 1 if z >= 0 else 0

def perceptron(X, y, lr=0.1, epochs=10):
    W = np.zeros(X.shape[1])
    b = 0

    for _ in range(epochs):
        for i in range(len(X)):
            z = np.dot(X[i], W) + b
            y_pred = step(z)
            W += lr * (y[i] - y_pred) * X[i]
            b += lr * (y[i] - y_pred)

    return W, b


X = np.array([[0,0], [0,1], [1,0], [1,1]])
y = np.array([0, 0, 0, 1])


W, b = perceptron(X, y)


predictions = [step(np.dot(x, W) + b) for x in X]
print(predictions)
```

```python
import numpy as np
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense
from tensorflow.keras.optimizers import Adam


X_train = np.random.rand(1000, 100, 1)
y_train = np.random.rand(1000, 1)
X_val = np.random.rand(200, 100, 1)
y_val = np.random.rand(200, 1)


optimizer = Adam(learning_rate=0.001, clipnorm=1.0)


model = Sequential([
    LSTM(50, return_sequences=True, input_shape=(100, 1)),
    LSTM(50),
    Dense(1, activation='linear')
])


model.compile(optimizer=optimizer, loss='mse')

model.fit(X_train, y_train, epochs=10, batch_size=32, validation_data=(X_val, y_val))
```

Q. 10 -  How can you create a custom loss function in Keras?

```python

def custom_huber_loss(y_true, y_pred, delta=1.0):
    error = y_true - y_pred
    is_small_error = tf.abs(error) <= delta
    squared_loss = 0.5 * tf.square(error)
    linear_loss = delta * (tf.abs(error) - 0.5 * delta)
    return tf.reduce_mean(tf.where(is_small_error, squared_loss, linear_loss))


X_train = np.random.rand(1000, 10)
y_train = np.random.rand(1000, 1)
X_val = np.random.rand(200, 10)
y_val = np.random.rand(200, 1)


model = Sequential([
    Dense(32, activation='relu', input_shape=(10,)),
    Dense(16, activation='relu'),
    Dense(1, activation='linear')
])


model.compile(optimizer=Adam(learning_rate=0.01), loss=lambda y_true, y_pred: custom_huber_loss(y_true, y_pred, delta=1.0))


model.fit(X_train, y_train, epochs=10, batch_size=32, validation_data=(X_val, y_val))

loss = model.evaluate(X_val, y_val)
print(f"Validation Loss: {loss}")
```
