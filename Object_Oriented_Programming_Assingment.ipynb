```python
# Q. 10 - Write a code that generates prime numbers up to n using a generator.

# Answer -

def generate_primes(n):

    def is_prime(num):
        if num < 2:
            return False
        for i in range(2, int(num ** 0.5) + 1):
            if num % i == 0:
                return False
        return True

    for i in range(2, n + 1):
        if is_prime(i):
            yield i
n = 50
for prime in generate_primes(n):
    print(prime)
```

# Q.37 - What is abstraction in Python? How is it Implemented?
# Answer -
Abstraction in Python is the concept of hiding the complex implementation details of a system and exposing only the necessary and relevant features to the user. The goal is to provide a simplified interface to the user while still hiding the complexity of the underlying operations. This makes the code easier to use and maintain.

In Python, abstraction can be achieved using:

1.Abstract Classes
2.Abstract Methods

How Abstraction is Implemented in Python:
Python provides the abc (Abstract Base Class) module to implement abstraction. The main components for implementing abstraction are:

ABC (Abstract Base Class): A class from which other classes can inherit.
abstractmethod: A decorator used to declare methods that must be implemented by subclasses.


```python
# Example of abstraction method.

from abc import ABC, abstractmethod
class Animal(ABC):

    @abstractmethod
    def sound(self):
        pass

    @abstractmethod
    def move(self):
        pass

class Dog(Animal):
    def sound(self):
        return "Woof"

    def move(self):
        return "Dog runs"

# Subclass implementing abstract methods
class Bird(Animal):
    def sound(self):
        return "Chirp"

    def move(self):
        return "Bird flies"

# Create instances of Dog and Bird
dog = Dog()
print(dog.sound())
print(dog.move())

bird = Bird()
print(bird.sound())
print(bird.move())
```

# Q. 39 - How are abstract methods different from regular mathods in Python?
## Answer -
In Python, abstract methods and regular methods differ primarily in their definition, usage, and purpose within object-oriented programming.

*Abstract Methods: These are methods that are declared in an abstract class (using the abc module) but do not have an implementation in the abstract class. Instead, they only define the method signature (name and parameters). Abstract methods act as templates or contracts, requiring any subclass to provide its own implementation. Abstract methods ensure that subclasses follow a specific interface, guaranteeing uniformity across different subclass implementations. You cannot instantiate an abstract class with unimplemented abstract methods directly.

*Regular Methods: These are methods defined in regular classes with both a signature and an implementation. Regular methods contain the actual logic and behavior of the class and can be invoked on instances of the class. They may or may not be overridden by subclasses.



```python
# Example Abstract Methods
from abc import ABC, abstractmethod

class Animal(ABC):
    @abstractmethod
    def sound(self):
        pass

# Example - Regular Methods:
class Dog:
    def sound(self):
        return "Woof"  # Full implementation

```

# Q. 40 -How can you achieve abstraction using interfaces in Python?

##Answer -

In Python, abstraction using interfaces can be achieved through Abstract Base Classes (ABCs) from the abc module. While Python does not have a formal interface keyword like Java or C#, abstract base classes can function as interfaces by declaring abstract methods that define the structure or contract for subclasses. These abstract methods only define the method signatures without providing an implementation. Subclasses are required to implement these methods, thus achieving abstraction

Steps to Achieve Abstraction Using Interfaces:
Define an Abstract Base Class (ABC): You create an abstract class using the ABC base class from the abc module. Inside this class, you define methods using the @abstractmethod decorator. These methods act as the interface, specifying what methods should exist, but not how they are implemented.

Implement the Interface in Subclasses: Subclasses that inherit from the abstract class must implement all abstract methods. This ensures that the concrete classes adhere to the structure defined by the abstract class.

Instantiating Concrete Classes: You cannot instantiate the abstract base class itself. You can only create instances of the subclasses that provide concrete implementations of the abstract methods.

```python
# Q. 41 - Can you provide an example of how abstraction can be utilized to create a common interface for a group of related classes in Python?
# Answer -

from abc import ABC, abstractmethod
class Shape(ABC):

    @abstractmethod
    def area(self):
        pass  # Abstract method to calculate area

import math

class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return math.pi * (self.radius ** 2)  # Area of circle: πr²

class Rectangle(Shape):
    def __init__(self, width, height):
        self.width = width
        self.height = height

    def area(self):
        return self.width * self.height  # Area of rectangle: width * height


```

```python
def print_area(shape: Shape):
    print(f"Area: {shape.area()}")

# Create instances of different shapes
circle = Circle(5)
rectangle = Rectangle(4, 6)

print_area(circle)
print_area(rectangle)
```

# Q. 42 - How does Python achieve polymorphism through method averriding?
## Answer -
Polymorphism Works in Python through Method Overriding: -
Method Overriding: In Python, polymorphism is typically achieved by method overriding, where a method in a subclass replaces or overrides the method defined in its superclass. This allows the subclass to provide its own implementation of the method while still maintaining the same method signature as the superclass.

Dynamic Method Resolution:-
When a method is called on an object, Python uses dynamic method resolution to determine which method to execute. It first looks for the method in the object's class, then in its parent classes (if any), until it finds the appropriate method.

Same Method Name, Different Implementations: The key to polymorphism in this context is that subclasses can have their own specific behavior for a method with the same name, and the method that gets called depends on the actual object type, not the reference type.

Example:-
Demonstrating Polymorphism with Method Overriding
Consider a simple example where we have a superclass Animal with a method make_sound. Different subclasses, such as Dog and Cat, override this method to provide their own specific implementations.

```python
# example -
class Animal:
    def make_sound(self):
        return "Some generic animal sound"

class Dog(Animal):
    def make_sound(self):
        return "Woof"

class Cat(Animal):
    def make_sound(self):
        return "Meow"

dog = Dog()
cat = Cat()
def animal_sound(animal: Animal):
    print(animal.make_sound())

animal_sound(dog)  # Output: Woof
animal_sound(cat)  # Output: Meow
```

# Q. 46 - Describe how Python supports polymorphism with duck typing.

## Answer -

Python supports polymorphism through duck typing, a programming principle that focuses on an object’s behavior rather than its class or type. The term comes from the saying: “If it looks like a duck, swims like a duck, and quacks like a duck, it probably is a duck.” In Python, this means that as long as an object implements the required methods or attributes, it can be used interchangeably with other objects, regardless of their type or inheritance.

Duck typing enables polymorphism by allowing functions or methods to work with any object that provides the expected behavior. Unlike traditional polymorphism, which often requires a common interface or base class, Python dynamically checks at runtime whether an object supports the necessary operations. This allows unrelated classes to exhibit polymorphic behavior.

```python
# Example -
class Dog:
    def speak(self):
        return "Woof!"

class Cat:
    def speak(self):
        return "Meow!"

def animal_sound(animal):
    print(animal.speak())

animal_sound(Dog())  # Output: Woof!
animal_sound(Cat())  # Output: Meow!
```

# Q. 47 - How do you achieve encapsulation in Python?
## Answer -
Public Members:- Attributes and methods declared without any prefix are public and accessible from outside the class.

Protected Members:- Prefixing an attribute or method with a single underscore (e.g., _age) makes it protected. These members are meant for internal use but can still be accessed from outside the class as a convention

Private Members:- Prefixing with double underscores (e.g., __salary) makes them private, restricting access to within the class. Private members are name-mangled by Python, making it harder to access them directly.

```python
class Employee:
    def __init__(self, name, salary):
        self.name = name           # Public
        self._position = "Staff"   # Protected
        self.__salary = salary     # Private

    def get_salary(self):
        return self.__salary       # Controlled access

emp = Employee("John", 50000)
print(emp.name)
print(emp._position)
print(emp.__salary)
print(emp.get_salary())
```

# Q. 48 - Can encapsulation be bypassed in Python? If so, how?
# Answer -
encapsulation can be bypassed in Python because it relies on conventions rather than strict enforcement. Protected members (prefixed with _) and private members (prefixed with __) can still be accessed outside their class.

Accessing Protected Members:

Protected attributes (e.g., _protected_attr) are only a convention to signal that they are for internal use but can be directly accessed or modified.

Accessing Private Members via Name Mangling:

Private attributes are name-mangled to prevent direct access, but their mangled name (_ClassName__attr) can be used.

# Q.68- Explain the role of setter methods in Python. Demonstrate how to use a setter method to modify a class attribute using property decorators.
## Answer -
In Python, a setter method is used to modify or set the value of an attribute in a controlled manner. Setter methods are especially useful when you need to enforce validation or other logic before modifying an attribute.

Purpose of Setter Methods:
Encapsulation: It helps to maintain the integrity of the object by ensuring that the attribute is modified only in a valid way.
Validation: You can add logic to check if the new value meets certain criteria before updating the attribute.
Control: Allows control over how an attribute is set (e.g., applying transformations or triggering side effects).

```python
class Person:
    def __init__(self, name, age):
        self._name = name
        self._age = age

    @property
    def age(self):
        return self._age

    @age.setter
    def age(self, value):
        if value < 0:
            raise ValueError("Age cannot be negative.")
        self._age = value

person = Person("Alice", 30)

person.age = 35  # This will call the setter

print(person.age)

```

# Q. 70 - Explain the use of the @deleter decorator in Python property decorators. Provide a code example demonstrating its application.

# Answer -

The deleter decorator in Python is used to define a deleter method for an attribute managed by the @property decorator. This allows you to define custom behavior when an attribute is deleted using the del statement.

Purpose of deleter:
The deleter decorator allows you to customize the deletion behavior of an attribute.
It is useful when you want to add logic for cleaning up resources or enforcing rules when an attribute is deleted.

```python
# Example -

class Rectangle:
    def __init__(self, width, height):
        self._width = width
        self._height = height

    @property
    def area(self):
        return self._width * self._height

    @property
    def width(self):
        return self._width

    @width.setter
    def width(self, value):
        if value <= 0:
            raise ValueError("Width must be positive")
        self._width = value

    @property
    def height(self):
        return self._height

    @height.setter
    def height(self, value):
        if value <= 0:
            raise ValueError("Height must be positive")
        self._height = value

    @property
    def dimensions(self):
        return f"Width: {self._width}, Height: {self._height}"

    @dimensions.deleter
    def dimensions(self):
        print("Deleting dimensions...")
        del self._width
        del self._height

rect = Rectangle(5, 10)
print(rect.area)
print(rect.dimensions)

del rect.dimensions

```
