Q 5. Generate a synthetic dataset with correlated features. Visualize the correlation matrix of a dataset using a heatmap.

```python
# Ans -
import seaborn as sns

np.random.seed(42)
size = 100  # Number of samples
feature1 = np.random.randn(size) * 10
feature2 = feature1 * 0.8 + np.random.randn(size) * 5
feature3 = np.random.randn(size) * 10
feature4 = feature1 * -0.5 + np.random.randn(size) * 3
import pandas as pd

data = pd.DataFrame({
    'Feature 1': feature1,
    'Feature 2': feature2,
    'Feature 3': feature3,
    'Feature 4': feature4
})


correlation_matrix = data.corr()

plt.figure(figsize=(8, 6))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt=".2f", linewidths=0.5)
plt.title('Correlation Matrix Heatmap')

plt.show()
```

```python
#Ans -
np.random.seed(25)

data = {
    'Country': ['USA', 'Canada', 'UK', 'Germany', 'France'],
    'Population': np.random.randint(100, 1000, 5),
    'GDP': np.random.randint(500, 2000, 5)
}

df = pd.DataFrame(data)

plt.figure(figsize=(8, 5))
plt.scatter(df['GDP'], df['Population'], s=df['Population']*2, c=df['Population'], cmap='coolwarm', alpha=0.6, edgecolors='black')


for i, country in enumerate(df['Country']):
    plt.text(df['GDP'][i], df['Population'][i], country, fontsize=12, ha='right')

plt.xlabel('GDP (in billion $)')
plt.ylabel('Population (in millions)')
plt.title('Bubble Chart of Population vs GDP')
plt.show()

```

# BOKEH ASSIGNMENT

Q 1.Create a Bokeh plot displaying a sine wave. Set x-values from 0 to 10 and y-values as the sine of x.

```python
# Ans -
from bokeh.plotting import figure, show
from bokeh.io import output_notebook
import numpy as np

output_notebook()
x = np.linspace(0, 10, 100)
y = np.sin(x)

p = figure(title="Sine Wave", x_axis_label='X', y_axis_label='sin(X)', width=700, height=400)
p.line(x, y, line_width=2, color='blue', legend_label="sin(x)")
show(p)
```

Q 2. Create a Bokeh scatter plot using randomly generated x and y values. Use different sizes and colors for the markers based on the 'sizes' and 'colors' columns.

```python
# Ans -
from bokeh.plotting import figure, show
from bokeh.io import output_notebook
from bokeh.models import ColumnDataSource
import numpy as np

output_notebook()


np.random.seed(42)
num_points = 100
x = np.random.rand(num_points) * 10
y = np.random.rand(num_points) * 10
sizes = np.random.randint(5, 20, num_points)
colors = np.random.randint(50, 255, num_points)


source = ColumnDataSource(data={'x': x, 'y': y, 'sizes': sizes, 'colors': colors})


p = figure(title="Bokeh Scatter Plot", x_axis_label="X Values", y_axis_label="Y Values", width=700, height=400)
p.scatter('x', 'y', size='sizes', color='colors', alpha=0.6, source=source, legend_label="Random Points")
show(p)
```

Q 3. Generate a Bokeh bar chart representing the counts of different fruits using the following dataset.

fruits = ['Apples', 'Oranges', 'Bananas', 'Pears'] counts [20, 25, 30, 35]


```python
#Ans -
from bokeh.plotting import figure, show
from bokeh.io import output_notebook

output_notebook()

fruits = ['Apples', 'Oranges', 'Bananas', 'Pears']
counts = [20, 25, 30, 35]

p = figure(x_range=fruits, title="Fruit Counts", x_axis_label="Fruit", y_axis_label="Count", width=700, height=400)

p.vbar(x=fruits, top=counts, width=0.5, color=['red', 'orange', 'yellow', 'green'])
show(p)

```

Q 4. Create a Bokeh histogram to visualize the distribution of the given data.

data_hist = np.random.randn(1000) hist, edges np.histogram(data_hist, bins=30)


```python
# Ans -

from bokeh.models import ColumnDataSource


data_hist = np.random.randn(1000)
hist, edges = np.histogram(data_hist, bins=30)


source = ColumnDataSource(data={'hist': hist, 'left': edges[:-1], 'right': edges[1:]})


p = figure(title="Histogram of Data Distribution", x_axis_label="Value", y_axis_label="Frequency", width=700, height=400)
p.quad(top='hist', bottom=0, left='left', right='right', source=source, fill_color="skyblue", line_color="black")
show(p)
```
