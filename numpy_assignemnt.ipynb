Q 44 -
For the given array, apply custom function to calculate moving sum along rows.

arr = np.array([[1, 2, 3], [4, 5, 6]])

```python
#Ans -
arr = np.array([[1, 2, 3],
                [4, 5, 6]])

def moving_sum(arr, window_size=2):
    result = np.array([np.convolve(row, np.ones(window_size, dtype=int), mode='valid') for row in arr])
    return result

result = moving_sum(arr, window_size=2)

print( result)

```

Q. 50 -
Write a function that inserts and then deletes elements from a given array at specified indices.

original_array = np.array([1, 2, 3, 4, 5])

indices_to_insert = [2, 4]

values_to_insert = [10, 11]

indices_to_delete = [1, 3]


```python
# Ans -
import numpy as np

def modify_array(original_array, indices_to_insert, values_to_insert, indices_to_delete):

    for idx, value in zip(indices_to_insert, values_to_insert):
        original_array = np.insert(original_array, idx, value)

    for idx in sorted(indices_to_delete, reverse=True):
        original_array = np.delete(original_array, idx)

    return original_array

original_array = np.array([1, 2, 3, 4, 5])
indices_to_insert = [2, 4]
values_to_insert = [10, 11]
indices_to_delete = [1, 3]

modified_array = modify_array(original_array, indices_to_insert, values_to_insert, indices_to_delete)
print(modified_array)
```

Q. 57 -
Write a function that extracts uppercase characters from a NumPy array of strings.

arr = np.array(['Hello', 'World', 'OpenAl', 'GPT'])

```python
# Ans -

import numpy as np
import re

def extract_uppercase(arr):
    return np.vectorize(lambda x: ''.join(re.findall(r'[A-Z]', x)))(arr)

arr = np.array(['Hello', 'World', 'OpenAI', 'GPT'])

uppercase_chars = extract_uppercase(arr)
print(f"Uppercase characters: {uppercase_chars}")

```

Q. 74 -
    For the given NumPy array and swap its byte order conditionally based on system endianness using `newbyteorder().

    arr = np.array([1, 2, 3], dtype=np.int32)

```python
# Ans -
import sys
arr = np.array([1, 2, 3], dtype=np.int32)


if sys.byteorder == 'little':

    swapped_arr = arr.newbyteorder('>')
else:

    swapped_arr = arr

print(arr)
print(swapped_arr)
```

Q. 75 -
For the given NumPy array, check if byte swapping is necessary for the current system using 'dtype" attribute byteorder`.

arr = np.array([1, 2, 3], dtype=np.int32)


```python
# Ans -
import sys

arr = np.array([1, 2, 3], dtype=np.int32)
array_byte_order = arr.dtype.byteorder

system_byte_order = sys.byteorder

if array_byte_order != system_byte_order:
    print("Byte swapping is necessary.")
else:
    print("Byte swapping is not necessary.")

print(f"Array byte order: {array_byte_order}")
print(f"System byte order: {system_byte_order}")
```
